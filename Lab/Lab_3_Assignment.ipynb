# Part A — Understand the Polynomial Regression Model

Ordinary Multiple Linear Regression assumes:

$$
[
\hat y
=
\beta_0
+
\beta_1TV
+
\beta_2Radio
+
\beta_3Newspaper
]
$$

However, the relationship between advertising expenditure and sales may not be perfectly linear.

For example:

- increasing TV advertising may initially increase sales rapidly,
- after a certain advertising level, the effect may change,
- TV and Radio advertising may also work together.

To capture such relationships, create a **second-degree Polynomial Regression model**:

$$
[
\begin{aligned}

+
\beta_8(TV \times Newspaper)

]
$$

The model contains:

- 1 intercept
- 3 original features

- 3 interaction features
